```python
## Q.1 Prime numbers
```

```python
num=int(input("Enter a number: "))  #input from user
if num > 1:
   for i in range(2,num):
     if (num % i)==0:                                             #applying the condition
      print(num,"is not a prime number")
      break
     else:
        print(num,"is a prime number")
        break
else:
    print(num,"is not a prime number")
```

```python
## Q.2 Product of Random Numbers
```

```python
import random                                                        #importing library
num1=random.randint(1,5)                                             #using the function randint from random library
num2=random.randint(5,10)
print(num1,num2)
answer=int(input(f"What is the product of {num1} & {num2}: "))
print(answer)
if answer==num1*num2:                                                 #applying the condition
  print("Your answer is correct")
else:
  print(f"Your answer is incorrect,Correct answer is {num1 * num2}")
```

```python
##Q.3 Squares of Even/Odd Numbers
```

```python
for num in range(100,201):                                        #using 'for' loop for iteration
  if num % 2 ==0:                                                 #applying the condition
    print(num,"square is",num ** 2)
```

```python
#Odd numbers:
for num in range(100,201):
  if num % 2 !=0:
    print(num,"square is",num ** 2)
```

```python
## Q.4 Word counter
```

```python
text="Hello from Excelr , Bye from Excelr"
words= text.split()
word_count={}
for word in words:
  if word in word_count:                                  #applying condition to check the count of number
    word_count[word]+=1
  else:
    word_count[word]=1
for word,count in word_count.items():
  print(f"'{word}':, {count}")
```

```python
## Q.5 Palindrome sequence
```

```python
def in_palindrome(text):                                          # defining 'is_palindrome' function
  text=text.replace(" ","").lower()
  return text == text[::-1]

a=input("Enter word/sequence:")
if in_palindrome(a):                                              #applying condition to check wheather given input is palindrome or not
  print(f"'{a}' is palindrome")
else:
  print(f"'{a}' is not palindrome")
```
