```python
Compute and analyze basic statistical measures for numerical columns in the dataset.
Calculate the mean, median, mode, and standard deviation for these columns
```

```python
import pandas as pd
df = pd.read_csv('/content/sales_data_with_discounts.csv')
#df.info()                                                   #identify numerical columns

"""
in the dataset volume,Avg Price,Total sales value,discount rate,discount amount and net sales value
are the numerical columns
"""

numerical_col = df.select_dtypes(include=['float64','int64'])
#print(numerical_col)

#calculate mean
mean = numerical_col.mean()
print("Mean for the numerical columns is:\n",mean)
print('********************************')

#calculate median
median = numerical_col.median()
print("Median for the numerical columns is:\n",median)
print('********************************')


#calculate mode
mode = numerical_col.mode()
print("Mode for the numerical columns is:\n",mode)
print('********************************')

#calculate stanmard deviation
std_dev = numerical_col.std()
print("Standard deviation for the numerical columns is:\n",std_dev)
print('********************************')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

data=pd.read_csv("/content/sales_data_with_discounts.csv")
numerical_cols = ['Volume','Avg Price','Total Sales Value','Discount Rate (%)','Discount Amount','Net Sales Value']

# HISTOGRAM:
for col in numerical_cols:
    plt.figure()
    plt.hist(data[col].dropna())
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

# BOXPLOT:5
for col in numerical_cols:
    plt.figure()
    plt.boxplot(data[col].dropna())
    plt.title(f"Boxplot of {col}")
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()

# Bar chart for categorical data:
categorical_cols = ['Date','Day','SKU','City','BU','Brand','Model']
for col in categorical_cols:
    plt.figure()
    data[col].value_counts().plot(kind='bar')
    plt.title(f"Bar Chart of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.tight_layout()
    plt.show()

# STANDARDIZATION
scaler = StandardScaler()
df_standardized = data.copy()
df_standardized[numerical_cols] = scaler.fit_transform(df[numerical_cols])

# Histogram:

for col in numerical_cols:
    plt.figure()
    plt.hist(df_standardized[col].dropna())
    plt.title(f"Histogram of {col} (Standardized)")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

df_scaled.head()

## one-hot encoding

df_encoded = pd.get_dummies(df_standardized, columns=categorical_cols, drop_first=True)

print("\nPreview of Encoded Dataset:\n")
print(df_encoded.head())
```
