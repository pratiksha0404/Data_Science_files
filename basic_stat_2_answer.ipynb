```python
import numpy as np
from scipy.stats import t, norm

# Data
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
n = len(data)
mean = np.mean(data)
std_sample = np.std(data, ddof=1)

# 99% Confidence Interval using t-distribution
confidence_level = 0.99
alpha = 1 - confidence_level
t_dist = t.ppf(1 - alpha/2, df=n-1)
error_t = t_dist * (std_sample / np.sqrt(n))
ci_t = (mean - error_t, mean + error_t)

# 99% Confidence Interval using known population standard deviation
sigma = 0.2
z_critical = norm.ppf(1 - alpha/2)
error_z = z_critical * (sigma / np.sqrt(n))
ci_z = (mean - error_z, mean + error_z)

(mean, std_sample, t_dist, ci_t, z_critical, ci_z)
```
